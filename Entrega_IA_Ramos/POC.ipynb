## Introducción  
-En este notebook se desarrolla una **Prueba de Concepto (POC)** para demostrar cómo aplicar **Fast Prompting** en el proyecto *Ambar Nails*.  
-El objetivo es documentar, analizar y generar ejemplos que muestren la utilidad de los prompts en la automatización de tareas, generación de ideas y análisis de información.

## Problema  
Los pequeños negocios como *Ambar Nails* necesitan materiales (texto e imágenes) que sean claros, visualmente atractivos y rápidos de producir.  
Crear este contenido manualmente es costoso y consume mucho tiempo, lo cual afecta la visibilidad y el alcance del emprendimiento.

## Objetivos  
-Aplicar técnicas de prompting para resolver problemas.  
-Analizar los resultados obtenidos.  
-Comparar la efectividad de diferentes tipos de prompts.  
-Documentar la experiencia en un notebook reproducible.  

## Metodología  
-Definir un conjunto de prompts de prueba.  
-Ejecutarlos y registrar los resultados.  
-Analizar la validez, claridad y utilidad de las respuestas.

# Simulación de prompts y resultados — versión Ambar Nails
prompts = [
    "Explicame qué es un esmalte de uñas en términos simples.",
    "Dame un ejemplo de cómo organizar un inventario de productos de uñas en una tabla.",
    "Genera un tip rápido de cuidado de uñas para redes sociales."
]

respuestas = [
    "Un esmalte de uñas es un líquido coloreado que se aplica sobre las uñas para decorarlas y protegerlas.",
    "Ejemplo de tabla de inventario: | Producto | Cantidad | Precio |\n| Esmalte rosa | 10 | $500 |\n| Lima de uñas | 5 | $200 |",
    "Tip rápido: Aplica siempre una capa base antes del esmalte para proteger tus uñas y que el color dure más."
]

for p, r in zip(prompts, respuestas):
    print(f"🔹 Prompt: {p}\n✅ Respuesta: {r}\n")
